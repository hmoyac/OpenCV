##  <font style="color:black">Download Assets</font>

The `download_and_unzip(...)` is used to download and extract the notebook assests.

````python
def download_and_unzip(url, save_path):
    print(f"Downloading and extracting assests....", end="")

    # Downloading zip file using urllib package.
    urlretrieve(url, save_path)

    try:
        # Extracting zip file using the zipfile package.
        with ZipFile(save_path) as z:
            # Extract ZIP file contents in the same directory.
            z.extractall(os.path.split(save_path)[0])

        print("Done")

    except Exception as e:
        print("\nInvalid file.", e)
````

## Accessing Individual Pixels
Let us see how to access a pixel in the image.

For accessing any pixel in a numpy matrix, you have to use matrix notation such as matrix[r,c], where the r is the row number and c is the column number. Also note that the matrix is 0-indexed.

For example, if you want to access the first pixel, you need to specify matrix[0,0]. Let us see with some examples. We will print one black pixel from top-left and one white pixel from top-center.

````python
# print the first pixel of the first black box
print(cb_img[0, 0])
# print the first white pixel to the right of the first black box
print(cb_img[0, 6])
````

# Resizing Images

The function **`resize`** resizes the image `src` down to or up to the specified size. The size and type are derived from the `src`,`dsize`,`fx`, and `fy`.

### <font style="color:rgb(8,133,37)">Function Syntax </font>

``` python
dst = resize( src, dsize[, dst[, fx[, fy[, interpolation]]]] )
```

`dst`: output image; it has the size dsize (when it is non-zero) or the size computed from src.size(), fx, and fy; the type of dst is the same as of src.

The function has **2 required arguments**:
    
1. `src`:	input image

2. `dsize`: output image size

Optional arguments that are often used include:

1. `fx`: Scale factor along the horizontal axis; when it equals 0, it is computed as (𝚍𝚘𝚞𝚋𝚕𝚎)𝚍𝚜𝚒𝚣𝚎.𝚠𝚒𝚍𝚝𝚑/𝚜𝚛𝚌.𝚌𝚘𝚕𝚜

2. `fy`: Scale factor along the vertical axis; when it equals 0, it is computed as (𝚍𝚘𝚞𝚋𝚕𝚎)𝚍𝚜𝚒𝚣𝚎.𝚑𝚎𝚒𝚐𝚑𝚝/𝚜𝚛𝚌.𝚛𝚘𝚠𝚜

The output image has the size `dsize` (when it is non-zero) or the size computed from `src.size()`, `fx`, and `fy`; the type of dst is the same as of src.

Interpolation Methods in OpenCV
OpenCV (cv2) offers several different interpolation methods. These interpolation methods are techniques used in image resizing, such as resolution enhancement or transformation processes. Here are some common cv2 interpolation methods:

1. cv2.INTER_NEAREST: Nearest-neighbor interpolation. It is the simplest and fastest method but can cause pixelation in the image. In this method, during resizing, the value of the new pixel is determined as the value of the nearest pixel in the original image.
<img src="./resize-interpolation-image/INTER_NEAREST.png">

2. cv2.INTER_LINEAR: Linear interpolation. It is usually the default method and produces smoother results. It is used when creating data between two points and is essentially calculated based on a straight-line equation.
<img src="./resize-interpolation-image/INTER_LINEAR.png">

3. cv2.INTER_CUBIC: Cubic interpolation. It uses 4x4 pixel neighbors to calculate and provides higher quality, but it is slower.
<img src="./resize-interpolation-image/INTER_CUBIC.png">

4. cv2.INTER_LANCZOS4: Lanczos interpolation. It uses 8x8 pixel neighbors and is one of the best for preserving details, but it works slowly. It creates new pixels during image enlargement or reduction by gathering information from surrounding pixels.
<img src="./resize-interpolation-image/INTER_LANCZOS4.png">

5. cv2.INTER_AREA: A method that causes less pixelation during resampling. It is often used when reducing image size and provides similar results to INTER_NEAREST when zooming in, but works in the opposite direction.
<img src="./resize-interpolation-image/INTER_AREA.png">

6. cv2.INTER_MAX: Automatically selects the highest quality method among all available interpolation techniques. However, its usage is rare.


````python
from datetime import datetime
plt.style.use('dark_background')
plt.figure(figsize=(28, 36))

# Resize image
a=[
    [cv2.INTER_NEAREST, "INTER_NEAREST"],
    [cv2.INTER_LINEAR, "INTER_LINEAR"],
    [cv2.INTER_CUBIC, "INTER_CUBIC"],
    [cv2.INTER_AREA, "INTER_AREA"],
    [cv2.INTER_LANCZOS4, "INTER_LANCZOS4"],
    [cv2.INTER_LINEAR_EXACT, "INTER_LINEAR_EXACT"],
    [cv2.INTER_NEAREST_EXACT, "INTER_NEAREST_EXACT"],
]

for interp in a:
  tini=datetime.now()
  resized_cropped_region = cv2.resize(img_NZ_rgb, dsize=None, fx=0.5, fy=0.5, interpolation=interp[0])
  tfin=datetime.now()
  tdiff=tfin-tini

  posi=a.index(interp)
  maxa= len(a)
  #print(posi)
  #posimg=int(str(1)+str(maxa)+str(posi+1))
  posimg=int(str(maxa)+str(1)+str(posi+1))
  #print(posimg)
  plt.subplot(posimg, xlabel='Width', ylabel='Height' );plt.imshow(resized_cropped_region);plt.title(interp[1]+ " - "+str(tdiff.microseconds)+" msec")
````

# Flipping Images


The function **`flip`** flips the array in one of three different ways (row and column indices are 0-based):

### <font style="color:rgb(8,133,37)">Function Syntax </font>
``` python
dst = cv.flip( src, flipCode )
```

`dst`: output array of the same size and type as src.

The function has **2 required arguments**:
    
1. `src`:	input image
2. `flipCode`: a flag to specify how to flip the array; 0 means flipping around the x-axis and positive value (for example, 1) means flipping around y-axis. Negative value (for example, -1) means flipping around both axes.


### <font style="color:rgb(8,133,37)">OpenCV Documentation</font>

**`flip`**: <a href="https://docs.opencv.org/4.5.0/d2/de8/group__core__array.html#gaca7be533e3dac7feb70fc60635adf441" target="_blank">Documentation link</a>

````python
img_NZ_rgb_flipped_horz = cv2.flip(img_NZ_rgb, 1)
img_NZ_rgb_flipped_vert = cv2.flip(img_NZ_rgb, 0)
img_NZ_rgb_flipped_both = cv2.flip(img_NZ_rgb, -1)

# Show the images
plt.figure(figsize=(18, 5))
plt.subplot(141);plt.imshow(img_NZ_rgb_flipped_horz);plt.title("Horizontal Flip");
plt.subplot(142);plt.imshow(img_NZ_rgb_flipped_vert);plt.title("Vertical Flip");
plt.subplot(143);plt.imshow(img_NZ_rgb_flipped_both);plt.title("Both Flipped");
plt.subplot(144);plt.imshow(img_NZ_rgb);plt.title("Original");
````
