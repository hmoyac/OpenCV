# Lab 1: Drawing Functions

OpenCV's drawing functions to create and annotate images, inspired by the OpenCV documentation but with new examples tailored for educational purposes.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Exercise 1: Drawing Shapes on a Blank Canvas

```python

# Create a blank 512x512 image
img = np.zeros((512, 512, 3), dtype=np.uint8)

# Draw a red line
cv2.line(img, (50, 50), (200, 200), (0, 0, 255), 2)

# Draw a green rectangle
cv2.rectangle(img, (100, 300), (300, 400), (0, 255, 0), 3)

# Draw a blue filled circle
cv2.circle(img, (400, 100), 50, (255, 0, 0), -1)

# Writing on Images
cv2.putText(img, 'Saddam Umer', (210, 230), cv2.FONT_HERSHEY_SIMPLEX, 1, (100, 210, 255), 3)


# Display and save
plt.imshow(img)
```

## Exercise 2: Annotating an Image

```python
# Load image
img = cv2.imread('/kaggle/input/sadams-practice-images-for-computer-vision/car.png')
if img is None:
    print("Error: Image not found.")
    exit()

# Draw yellow rectangle
cv2.rectangle(img, (100, 150), (670, 400), (0, 0, 255), 2)

# Add text
cv2.putText(img, 'Car', (100, 140), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

# Display and save
plt.imshow(img)
```

## Exercise 3: Circle Grid Pattern

```python

# Create white 512x512 image
img = np.ones((512, 512, 3), dtype=np.uint8) * 255

# Draw 5x5 grid of circles
for i in range(5):
    for j in range(5):
        center = (100 + j * 80, 100 + i * 80)
        color = (0, 0, 255) if (i + j) % 2 == 0 else (255, 0, 0)
        cv2.circle(img, center, 20, color, -1)

# Display and save
plt.imshow(img)
```
