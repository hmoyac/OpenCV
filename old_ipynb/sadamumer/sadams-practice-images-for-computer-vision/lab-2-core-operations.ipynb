# Lab 2: Core Operations in OpenCV

This lab focuses on core image processing operations using OpenCV, including pixel manipulation, image arithmetic, and region of interest (ROI) handling. The code is designed to run in a Jupyter Notebook, with images displayed inline using matplotlib instead of OpenCV windows.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## Exercise 1: Pixel Manipulation

Task: Load an image, modify pixel values to create a grayscale effect manually, and display it inline.

```python
# Load image
img = cv2.imread('car.png')
if img is None:
    print("Error: Image not found.")
    exit()

# Manual grayscale conversion
gray_img = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        # Average RGB values
        gray_img[i, j] = int(np.mean(img[i, j]))

# Display images
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title('Manual Grayscale')
plt.imshow(gray_img, cmap='gray')
plt.axis('off')
```

## Exercise 2: Image Arithmetic

Task: Perform addition and subtraction on two images (or the same image with itself) and display results.

```python
# Load image
img = cv2.imread('car.png')
if img is None:
    print("Error: Image not found.")
    exit()

# Create a brighter version by adding a constant
bright_img = cv2.add(img, np.array([50, 50, 50], dtype=np.uint8))

# Create a darker version by subtracting a constant
dark_img = cv2.subtract(img, np.array([50, 50, 50], dtype=np.uint8))

# Display images
plt.figure(figsize=(20, 5))
plt.subplot(1, 3, 1)
plt.title('Original Image')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.subplot(1, 3, 2)
plt.title('Brightened Image')
plt.imshow(cv2.cvtColor(bright_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.subplot(1, 3, 3)
plt.title('Darkened Image')
plt.imshow(cv2.cvtColor(dark_img, cv2.COLOR_BGR2RGB))
plt.axis('off')
```

## Exercise 3: Region of Interest (ROI)

Task: Extract a region of interest (e.g., part of the car) and paste it elsewhere in the image.

```python
# Load image
img = cv2.imread('/kaggle/input/sadams-practice-images-for-computer-vision/car.png')
if img is None:
    print("Error: Image not found.")
    exit()

# Define ROI (adjust coordinates based on your image)
roi = img[240:350, 90:150]  # Example: part of the car

# Paste ROI to another location
img[40:150, 90:150] = roi

# Display result
plt.figure(figsize=(8, 6))
plt.title('Image with Pasted ROI')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
# plt.axis('off')
```
