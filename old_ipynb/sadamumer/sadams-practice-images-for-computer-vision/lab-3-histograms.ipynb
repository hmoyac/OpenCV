# Lab 3: Histograms in OpenCV
This lab focuses on computing, visualizing, and manipulating image histograms using OpenCV. The exercises cover histogram calculation, visualization, and equalization, with code designed to run in a Jupyter Notebook, displaying results inline using matplotlib.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## Exercise 1: Computing and Visualizing a Grayscale Histogram

Task: Load an image, compute its grayscale histogram, and visualize it.

```python
# Load image and convert to grayscale
img = cv2.imread('/kaggle/input/sadams-practice-images-for-computer-vision/car.png', cv2.IMREAD_GRAYSCALE)
if img is None:
    print("Error: Image not found.")
    exit()

# Compute histogram
hist = cv2.calcHist([img], [0], None, [256], [0, 256])

# Display image and histogram
plt.figure(figsize=(20, 5))
plt.subplot(1, 2, 1)
plt.title('Grayscale Image')
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title('Grayscale Histogram')
plt.plot(hist, color='black')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True)
```

## Exercise 2: Color Histogram for RGB Channels

Task: Compute and visualize histograms for each RGB channel of a color image.

```python
# Load color image
img = cv2.imread('/kaggle/input/sadams-practice-images-for-computer-vision/car.png')
if img is None:
    print("Error: Image not found.")
    exit()

# Compute histograms for each channel
colors = ('b', 'g', 'r')
plt.figure(figsize=(10, 5))
plt.title('RGB Histograms')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True)

for i, color in enumerate(colors):
    hist = cv2.calcHist([img], [i], None, [256], [0, 256])
    plt.plot(hist, color=color, label=f'{color.upper()} Channel')

plt.legend()

# Display original image
plt.figure(figsize=(10, 5))
plt.title('Original Image')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
```

## Exercise 3: Histogram Equalization

Task: Apply histogram equalization to a grayscale image and compare histograms before and after.

```python
# Load image in grayscale
img = cv2.imread('/kaggle/input/sadams-practice-images-for-computer-vision/car.png', cv2.IMREAD_GRAYSCALE)
if img is None:
    print("Error: Image not found.")
    exit()

# Apply histogram equalization
equ_img = cv2.equalizeHist(img)

# Compute histograms
hist_orig = cv2.calcHist([img], [0], None, [256], [0, 256])
hist_equ = cv2.calcHist([equ_img], [0], None, [256], [0, 256])

# Display images and histograms
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.title('Original Image')
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.title('Equalized Image')
plt.imshow(equ_img, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.title('Original Histogram')
plt.plot(hist_orig, color='black')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True)

plt.subplot(2, 2, 4)
plt.title('Equalized Histogram')
plt.plot(hist_equ, color='black')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')
plt.grid(True)

plt.tight_layout()
```
