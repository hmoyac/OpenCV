```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Load two images
img1 = cv2.imread('cervante/img/Cervante-logo.jpg')
img2 = cv2.imread('cervante/img/dragonskull.png')

desired_width = 100
aspect_ratio = desired_width / img1.shape[1]
desired_height = int(img1.shape[0] * aspect_ratio)
dim = (desired_width, desired_height)

img2 = cv2.resize(img2, dim, interpolation=cv2.INTER_AREA)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
# # I want to put logo on top-left corner, So I create a ROI
rows, cols, channels = img2.shape

margen_left = 20
margen_right = 20
rx = img1.shape[0] - margen_left
ry = img1.shape[1] - margen_right
roi = img1[rx - desired_width:rx, ry - desired_height:ry]

# Now create a mask of logo and create its inverse mask also
img2gray = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
ret, mask = cv2.threshold(img2gray, 10, 255, cv2.THRESH_BINARY)
mask_inv = cv2.bitwise_not(mask)
# Now black-out the area of logo in ROI
img1_bg = cv2.bitwise_and(roi, roi, mask=mask_inv)
# Take only region of logo from logo image.
img2_fg = cv2.bitwise_and(img2, img2, mask=mask)

# Put logo in ROI and modify the main image
dst = cv2.add(img1_bg, img2_fg)
dst = cv2.cvtColor(dst, cv2.COLOR_BGR2RGB)
img1[rx - desired_width:rx, ry - desired_height:ry] = dst
#img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
cv2.imwrite("cervante/img/cervante-dragonskull.jpg", img1)
plt.figure(figsize=[20, 20])
plt.subplot(161);
plt.imshow(img2gray, cmap="gray")
plt.subplot(162);
plt.imshow(mask, cmap="gray")
plt.subplot(163);
plt.imshow(mask_inv, cmap="gray")
plt.subplot(164);
plt.imshow(img1_bg)
plt.subplot(165);
plt.imshow(img2_fg)
plt.subplot(166);
plt.imshow(dst)
plt.figure(figsize=[20, 20])
plt.subplot(111);
plt.imshow(img1[:, :, ::-1])
```
